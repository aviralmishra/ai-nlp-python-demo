```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
en_stopwords = stopwords.words('english')
```

```python
print(en_stopwords)
```

```python
sentence = 'I did not want to go to the movies because the tickets were too expensive.'
```

```python
x_sentence = ' '.join([word for word in sentence.split() if word not in en_stopwords])
```

```python
print(x_sentence)
```

```python
en_stopwords.remove('too')
```

```python
en_stopwords.append('I')
```
