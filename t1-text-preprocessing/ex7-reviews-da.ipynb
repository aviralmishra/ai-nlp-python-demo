```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import re
import pandas as pd
```

**Load**

```python
data = pd.read_csv('tripadvisor_hotel_reviews.csv')
```

```python
data.info()
```

```python
data.head()
```

```python
data['Review'][0]
```

**Lower case**

```python
data['review_lowercase'] = data['Review'].str.lower()
```

```python
data.head()
```

**Stop word removal**

```python
en_stopwords = stopwords.words('english')
en_stopwords.remove("not")
```

```python
data['review_no_stopwords'] = data['review_lowercase'] \
                                .apply(lambda x: ' '
                                .join([word for word in x.split() if word not in (en_stopwords)]))
```

```python
data['review_lowercase'][0]
```

```python
data['review_no_stopwords'][0]
```

**Punctuation**

```python
data['review_no_stopwords_or_punct'] = data \
                                        .apply(lambda x: re.sub(r"[*]", "star", x['review_no_stopwords']), 
                                        axis = 1)
```

```python
data['review_no_stopwords_or_punct'] = data \
                                    .apply(lambda x: re.sub(r"([^\w\s])", "", x['review_no_stopwords_or_punct']),
                                    axis=1)
```

```python
data['review_no_stopwords_or_punct'][2]
```

**Tokenizing**

```python
data['tokenized'] = data.apply(lambda x: \
                               word_tokenize( \
                               x['review_no_stopwords_or_punct'] \
                               ), axis = 1)
```

```python
data.head(1)
```

```python
data['tokenized'][0]
```

**Stemming**

```python
ps = PorterStemmer()
```

```python
data["stemmed"] = data["tokenized"] \
                  .apply(lambda tokens: [ps.stem(token) for token in tokens])
```

```python
data['stemmed'][0]
```

**Lemmatization**

```python
lemmatizer = WordNetLemmatizer()
```

```python
data["lemmatized"] = data["tokenized"] \
                    .apply(lambda tokens: [lemmatizer.lemmatize(token) for token in tokens])
```

```python
data['lemmatized'][0]
```

**N-grams**

```python
tokens_clean = sum(data['lemmatized'], [])
```

```python
unigrams = (pd.Series \
            (nltk.ngrams(tokens_clean, 1)) \
            .value_counts()) 
print(unigrams)
```

```python
bigrams = (pd.Series \
           (nltk.ngrams(tokens_clean, 2)) \
           .value_counts()) 
print(bigrams)
```

```python
ngrams_4 = (pd.Series \
            (nltk.ngrams(tokens_clean, 4)) \
            .value_counts()) 
print(ngrams_4)
```
